```python
# Step 1: Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Step 2: Create a sample dataset
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [200, 220, 250, 270, 300, 330],
    'Expenses': [180, 190, 210, 230, 250, 260]
}
df = pd.DataFrame(data)
df
```

```python
# Step 3: Quick statistics
df.describe()
```

```python
# Step 4: Plot Sales vs Expenses
plt.figure(figsize=(8,5))
plt.plot(df['Month'], df['Sales'], marker='o', label='Sales')
plt.plot(df['Month'], df['Expenses'], marker='o', label='Expenses')
plt.title('Company Performance - First Half Year')
plt.xlabel('Month')
plt.ylabel('Amount ($)')
plt.legend()
plt.grid(True)
plt.show()
```
