```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva","Ravi","Suchitra"],
    "Age": [25, 30, 35, 40, 45,24,35],
    "Department": ["HR", "Finance", "IT", "Marketing", "Finance","HR","IT"],
    "Salary": [50000, 60000, 75000, 80000, 65000,45000,60000]
}

df = pd.DataFrame(data)
df
```
